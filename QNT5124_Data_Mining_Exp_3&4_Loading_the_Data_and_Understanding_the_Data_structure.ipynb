### Understanding the Data structure

#### Titanic Dataset — Understanding the Data Structure

The following table describes each attribute in the Titanic dataset, its meaning, how Pandas stores it, and its statistical nature.

| No. | Column Name | Descriptive Meaning | Pandas Data Type | Meaning of Data Type | Statistical Nature |
|---:|---|---|---|---|---|
| 1 | **survived** | Whether the passenger survived: `0 = No`, `1 = Yes` | `int64` | Integer numeric value | **Binary** |
| 2 | **pclass** | Passenger ticket class: `1 = First`, `2 = Second`, `3 = Third` | `int64` | Integer numeric value | **Ordinal** |
| 3 | **sex** | Passenger's sex/gender category, e.g., male or female | `object` | Text/string data | **Nominal Categorical** |
| 4 | **age** | Age of the passenger in years | `float64` | Decimal/continuous numeric value | **Numerical — Continuous** |
| 5 | **sibsp** | Number of siblings or spouses travelling with the passenger | `int64` | Integer numeric count | **Numerical — Discrete** |
| 6 | **parch** | Number of parents or children travelling with the passenger | `int64` | Integer numeric count | **Numerical — Discrete** |
| 7 | **fare** | Passenger's ticket fare | `float64` | Decimal/continuous numeric value | **Numerical — Continuous** |
| 8 | **embarked** | Port where the passenger boarded: `C`, `Q`, or `S` | `object` | Text/string data | **Nominal Categorical** |
| 9 | **class** | Passenger class represented as First, Second, or Third | `category` | Pandas categorical data type | **Ordinal Categorical** |
| 10 | **who** | Broad passenger category, such as `man`, `woman`, or `child` | `object` | Text/string data | **Nominal Categorical** |
| 11 | **adult_male** | Whether the passenger was an adult male: `True` or `False` | `bool` | Boolean data type | **Binary** |
| 12 | **deck** | Deck on which the passenger's cabin was located | `category` | Pandas categorical data type | **Nominal Categorical** |
| 13 | **embark_town** | Name of the town/port where the passenger boarded | `object` | Text/string data | **Nominal Categorical** |
| 14 | **alive** | Whether the passenger was alive: `yes` or `no` | `object` | Text/string data | **Binary Categorical** |
| 15 | **alone** | Whether the passenger was travelling alone: `True` or `False` | `bool` | Boolean data type | **Binary** |

### Important Point

**Pandas data type ≠ Statistical nature of the attribute**

The Pandas `dtype` tells us **how Python stores the data**, whereas the statistical nature tells us **what the data represents**.

For example:

- `pclass` → stored as `int64`, but statistically it is **ordinal**.
- `survived` → stored as `int64`, but statistically it is **binary**.
- `age` → stored as `float64` and is **continuous numerical** data.
- `sex` → stored as `object`, but is **nominal categorical** data.
- `adult_male` → stored as `bool` and is **binary** data.
- `fare` → stored as `float64` and is **continuous numerical** data.

> **Key idea:** Never determine the statistical nature of an attribute from its Pandas `dtype` alone. Understand what the attribute actually represents.
